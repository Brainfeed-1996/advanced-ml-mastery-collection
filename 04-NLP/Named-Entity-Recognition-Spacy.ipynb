# Named Entity Recognition — spaCy

Loads spaCy + English model, runs NER, visualizes entities. Downloads model if missing.

_Last rebuild: **2026-02-16 03:36:21**_

```python
import spacy
from spacy.cli import download

model='en_core_web_sm'
try:
    nlp = spacy.load(model)
except Exception:
    download(model)
    nlp = spacy.load(model)

text = 'Olivier lives in Rennes and works on OpenClaw with GitHub. Apple released a new iPhone in Paris.'
doc = nlp(text)
[(ent.text, ent.label_) for ent in doc.ents]
```

```python
# Render to HTML (works in Jupyter).
# In nbconvert execution, displacy may return None depending on environment.
from spacy import displacy
html = displacy.render(doc, style='ent')
if isinstance(html, str):
    print(html[:500])
else:
    print('displacy.render returned:', type(html))
```
