```python
from ultralytics import YOLO
import numpy as np
from PIL import Image, ImageDraw

# Create a simple synthetic image (no external files)
img = Image.new('RGB', (640, 384), color=(30, 30, 30))
d = ImageDraw.Draw(img)
d.rectangle([200, 120, 440, 300], outline=(255, 255, 255), width=6)
img_path = 'synthetic_scene.jpg'
img.save(img_path)

model = YOLO('yolov8n.pt')
results = model(img_path, verbose=False)
results[0].boxes
```

```python
# Visualize detections
from matplotlib import pyplot as plt

im = results[0].plot()  # numpy array
plt.figure(figsize=(10,5))
plt.imshow(im)
plt.axis('off')
plt.show()
```
